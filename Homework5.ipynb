```python
import cv2
import numpy as np

# 책 표지 이미지 로드
img1 = cv2.imread("toeicbook.jpg", cv2.IMREAD_GRAYSCALE)

# SIFT 생성
sift = cv2.SIFT_create()
kp1, des1 = sift.detectAndCompute(img1, None)

# Brute Force Matcher 생성
bf = cv2.BFMatcher(cv2.NORM_L2, crossCheck=True)

# 카메라 열기
cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # 카메라 프레임 그레이스케일 변환
    img2 = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # 카메라 프레임에서 특징점 추출
    kp2, des2 = sift.detectAndCompute(img2, None)

    # 매칭 수행
    if des2 is not None and len(des2) > 0:
        matches = bf.match(des1, des2)
        matches = sorted(matches, key=lambda x: x.distance)

        # 매칭 결과 시각화
        matching_result = cv2.drawMatches(
            img1, kp1, img2, kp2, matches[:50], None, flags=2
        )
        cv2.imshow("Matching Result", matching_result)
    else:
        cv2.imshow("Matching Result", frame)

    # 'q' 키 누르면 종료
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
