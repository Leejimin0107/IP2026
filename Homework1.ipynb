```python
import cv2
import numpy as np
```

```python
org_img = cv2.imread('mountain.jpg')
img = org_img.copy()
```

```python
def nothing(x):
    pass
```

```python
drawing = False
ix, iy = -1, -1
cx, cy = -1, -1
final_rect = None  # 드래그 완료 후 남겨둘 사각형 좌표

org_img = cv2.imread('mountain.jpg')
img = org_img.copy()

# mouse callback function
def draw_circle(event, x, y, flags, param):
    global ix, iy, cx, cy, drawing, final_rect
    if event == cv2.EVENT_LBUTTONDOWN:
        drawing = True
        ix, iy = x, y
        cx, cy = x, y
    elif event == cv2.EVENT_MOUSEMOVE:
        if drawing:
            cx, cy = x, y
    elif event == cv2.EVENT_LBUTTONUP:
        drawing = False
        cx, cy = x, y
        final_rect = (ix, iy, cx, cy)  # 완료된 사각형 저장
```

```python
cv2.namedWindow('image')
cv2.createTrackbar('value', 'image', 80, 255, nothing)  # 기본 투명도
cv2.setMouseCallback('image', draw_circle)
font = cv2.FONT_HERSHEY_SIMPLEX

while(1):
    img = org_img.copy()
    value = cv2.getTrackbarPos('value', 'image')
    alpha = value / 255.0

    # 드래그 중에는 테두리만 미리보기
    if drawing:
        cv2.rectangle(img, (ix, iy), (cx, cy), (0, 0, 255), 2)

    # 드래그가 끝난 뒤에는 반투명 빨간 박스를 유지
    if final_rect is not None:
        x1, y1, x2, y2 = final_rect
        overlay = img.copy()
        cv2.rectangle(overlay, (x1, y1), (x2, y2), (0, 0, 255), -1)
        img = cv2.addWeighted(overlay, alpha, img, 1 - alpha, 0)

    cv2.putText(img, f"Mouse position : ({ix},{iy})-({cx},{cy})-{value}",
        (10,30), font, 0.7, (255,255,255), 2, cv2.LINE_AA)
    cv2.imshow('image', img)

    k = cv2.waitKey(1) & 0xFF
    if k == 27:
        break

cv2.destroyAllWindows()
```
